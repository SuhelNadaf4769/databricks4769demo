```python
import pandas as pd
import numpy as np

# Total rows to generate 
num_rows = 1000000

# To ensure duplicates, we create a pool of unique items smaller than the total rows
# Choosing 700000 unique items mean roughly 300000 rows will be duplicates
num_unique = 700000

#1. Customer table(5 columns)
# customer_id, name, email, city, membership_tier

cust_pool_indices = np.random.choice(num_unique, size = num_rows, replace = True)
unique_cust_ids = np.arange(10000, 10000 + num_unique)

customers = pd.DataFrame({
    "customer_id": unique_cust_ids[cust_pool_indices],
    "name": [f"User_{i}" for i in unique_cust_ids[cust_pool_indices]],
    "email": [f"user_{i}@example.com" for i in unique_cust_ids[cust_pool_indices]],
    "city": np.random.choice(["New York", "London", "Tokyo", "Paris", "Berlin"], size=num_rows),
    "membership_tier": np.random.choice(["Gold", "Silver", "Bronze"], size=num_rows)
})

# 2. Products Table (5 columns)
# product_id, product_name, category, price, stock
prod_pool_indices = np.random.choice(num_unique, size=num_rows, replace=True)
unique_prod_ids = np.arange(50000, 50000 + num_unique)

products = pd.DataFrame({
    "product_id": unique_prod_ids[prod_pool_indices],
    "product_name": [f"Item_{i}" for i in unique_prod_ids[prod_pool_indices]],
    "category": np.random.choice(["Electronics", "Home", "Fashion", "Beauty"], size=num_rows),
    "price": np.round(np.random.uniform(10, 500, size=num_rows), 2),
    "stock": np.random.randint(0, 1000, size=num_rows)
})

```

```python
customers_df = spark.createDataFrame(customers)
products_df = spark.createDataFrame(products)

display(customers_df)
display(products_df)
```

```python
customers_df.write.format("delta").option("mode", "overwrite").saveAsTable("customers_bronze")
products_df.write.format("delta").option("mode", "overwrite").saveAsTable("products_bronze")

```
